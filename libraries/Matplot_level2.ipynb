```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Sample Dateset

we will create a reatil store dataset used throughout the notebook

Columns:
- Product
- Sales
- Profit
- Advertising Budget
- Units Sold

```python
data = {
    "Product": ["Laptop","Phone","Tablet","Monitor","Keyboard","Mouse","Printer","Speaker"],
    "Sales": [1200,900,650,400,150,120,300,220],
    "Profit": [300,250,120,100,40,30,80,60],
    "Advertising": [200,180,120,90,40,35,70,60],
    "Units_Sold": [80,150,120,60,200,220,50,75]
}

df = pd.DataFrame(data)
df
```

# Bar Chart

## Types of Bar Charts
1. Horizontal Bar Chart
2. Vertical Bar Chart
3. Multiple Bar Chart


## Syntax

plt.bar(x, height)



## Vertical Bar Chart

```python

plt.bar(df["Product"],df["Sales"])

plt.title("Product Sales Comparison")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

## Horizontal Bar Chart

Used when category labels are long.


```python
plt.barh(df["Product"],df["Profit"])

plt.title("Profit by Product")
plt.xlabel("Profit")
plt.ylabel("Product")

plt.show()
```

## Multiple Bar Chart

Comparing Multilple metrics together.

```python
x = np.arange(len(df["Product"]))
width = 0.35

plt.bar(x, df["Sales"], width, label="Sales")
plt.bar(x + width, df["Profit"], width, label="Profit")

plt.xticks(x + width/2, df["Product"])

plt.title("Sales vs Profit by Product")
plt.xlabel("Product")
plt.ylabel("Amount")

plt.legend()

plt.show()
```
