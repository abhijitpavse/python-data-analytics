# Import Libraries

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

# Sample Dataset

We will use a **sales and marketing dataset**.

Columns:
- Month
- Sales
- Profit
- Marketing Spend
- Customer Visits


````python
data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Sales": [200,220,250,240,270,300,320,310,330,360,380,400],
    "Profit": [50,60,70,65,75,90,100,95,110,120,130,150],
    "Marketing": [40,45,50,48,55,60,65,63,70,75,80,90],
    "Visits": [500,520,580,560,600,650,680,670,700,720,760,800]
}

df = pd.DataFrame(data)
df
````

# 1. Subplots

Subplots allow you to **display multiple charts in a single figure**.

This is useful when:
- Comparing multiple metrics
- Building dashboards
- Showing trends side-by-side

## Syntax

```
plt.subplot(rows, columns, position)
```


## Basic Subplot Example

````python
plt.figure(figsize=(10,6))

plt.subplot(1,2,1)
plt.plot(df["Month"], df["Sales"])
plt.title("Monthly Sales")

plt.subplot(1,2,2)
plt.plot(df["Month"], df["Profit"])
plt.title("Monthly Profit")

plt.show()
````

## Using plt.subplots()

A cleaner and recommended approach.


````python
fig, ax = plt.subplots(2,1)

ax[0].plot(df["Month"], df["Sales"])
ax[0].set_title("Sales Trend")

ax[1].plot(df["Month"], df["Profit"])
ax[1].set_title("Profit Trend")

plt.tight_layout()
plt.show()
````

# 2. Axes Customization

Axes customization helps control how data appears.

Common operations:
- Setting axis limits
- Changing tick positions
- Changing tick labels
- Log scale visualization


## Axis Limits

Functions:
```
plt.xlim()
plt.ylim()
```


````python
plt.plot(df["Month"], df["Sales"])

plt.ylim(150,450)

plt.title("Sales with Custom Y Limits")

plt.show()
````

## Custom Ticks

Ticks represent values on axes.


````python
plt.plot(df["Month"], df["Sales"])

plt.xticks(rotation=45)

plt.title("Custom X Tick Rotation")

plt.show()
````

## Log Scale

Used when values vary greatly.


````python
large_values = [10,100,1000,10000,100000]

plt.plot(large_values)

plt.yscale("log")

plt.title("Log Scale Example")

plt.show()
````

# 3. Colormaps

Colormaps allow color variation based on data values.

Useful for:
- Heat intensity
- Density visualization
- Scientific data

Common colormaps:
- viridis
- plasma
- coolwarm


## Colormap Scatter Example

````python
np.random.seed(1)

x = np.random.rand(100)
y = np.random.rand(100)
colors = np.random.rand(100)

plt.scatter(x,y,c=colors,cmap="magma")

plt.title("Scatter Plot with Colormap")

plt.show()
````

## Colormap with Colorbar

Colorbar explains color meaning.


````python
scatter = plt.scatter(x,y,c=colors,cmap="plasma")

plt.colorbar(scatter)

plt.title("Scatter with Colorbar")

plt.show()
````

# 4. Annotations

Annotations highlight **important data points**.

Useful for:
- Explaining peaks
- Highlighting outliers
- Drawing attention to events

Functions:
```
plt.annotate()
plt.text()
```


## Annotating Maximum Sales

````python
max_sales = df["Sales"].max()
max_month = df["Month"][df["Sales"].idxmax()]

plt.plot(df["Month"], df["Sales"])

plt.annotate("Highest Sales",
             xy=(max_month, max_sales),
             xytext=(max_month, max_sales+20),
             arrowprops=dict(arrowstyle="|-|"))

plt.title("Sales with Annotation")

plt.show()
````

## Adding Text to Plot

````python
plt.plot(df["Month"], df["Profit"])

plt.text(5,120,"Profit Growth Area")

plt.title("Profit Trend with Text")

plt.show()
````
