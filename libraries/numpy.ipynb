```python
import numpy as np
# check version 
print("Numpy Version:",np.__version__)
```

```python
# python list example
my_list = [1, 2, 3, 4]
print("Python List:", my_list, " | Type",type(my_list))
print(f"Python List: {my_list} | Type: {type(my_list)}")

# add 5 to each element (needs a loop or comprehension)
list_result = [ x + 5 for x in my_list]
print("After adding 5:", list_result)
```

```python
# Numpy array example
my_array = np.array([1, 2, 3, 4])
print("Numpy Array: ",my_array, " | Type:", type(my_array))

# vectorized addition (no loop required)
array_result = my_array + 5
print("After adding 5:", array_result)
```

```python
# Speed comparison

import time

size = 10_000_000
# same as size = 10000000 10 million element
# python list
list1 = list(range(size))
start = time.time()
list2 = [x*2 for x in list1]
print("Python list time:", round(time.time() - start, 4),"sec")

# Numpy array
array1 = np.arange(size)
start = time.time()
array2 = array1 * 2
print("Numpy array time:", round(time.time() -start, 4),"sec")
```

```python
# 1D Array
a = np.array([1, 2, 3, 4, 5])
print(a)

# 2D Array
b = np.array([[1, 2, 3], [4, 5, 6]])
print(b)

# specify dtype
c = np.array([1, 2, 3], dtype=float)
print(c)
```

# (b) Using np.arange(start, stop, step)

```python
arr1 = np.arange(0, 10)
print("arr1:", arr1)

arr2 = np.arange(0, 20, 2)
print("arr2:", arr2)

arr3 = np.arange(0, 1, 0.1)
print("arr3:", arr3)
```

# (c) Using np.linspace(start, stop, num)

```python
lin1 = np.linspace(0, 20, 5)
print("lin1:", lin1)

lin2 = np.linspace(10, 50, 9)
print("lin2:", lin2)
```

# (d) Using np.zeros

```python
z1 = np.zeros(5)
print("z1:", z1)

z2 = np.zeros((2, 3))
print("z2:", z2)

z3 = np.zeros((3, 3), dtype=int)
print("z3:", z3)
```

# (e) Using np.ones

```python
o1 = np.ones(4)
print("o1:", o1)

o2 = np.ones((3, 2))
print("o2:", o2)

o3 = np.ones((2, 2), dtype=int)
print("o3:", o3)
```

# (f) Using `np.eye` (N) -- Identity matrix

```python
I = np.eye(5)
print("I:", I)

I2 = np.identity(3)
print("I2:", I2)
```
