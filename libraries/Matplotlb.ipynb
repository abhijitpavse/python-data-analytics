```python
import pandas as pd
data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Sales": [120,150,170,160,180,200,210,190,220,230,250,270],
    "Profit": [20,25,30,28,35,40,45,42,50,55,60,65]
}

df = pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt
```

```python
x= df["Month"]
y= df["Sales"]

plt.plot(x,y)
plt.show()
```

# Axes

- figure = entire page
- axes = chart inside the page

```python
fig, ax = plt.subplots()
ax.plot(df["Month"], df["Sales"])
plt.show()
```

## Creating multiple Axes
You can crate multiple charts in one figure. 

```python
fig, ax = plt.subplots(1,2)

ax[0].plot(df["Month"], df["Sales"])
ax[1].plot(df["Month"], df["Profit"])

plt.show()
```

# 5. Labels and Titles
labes make charts easier to understand

function used:
- `plt.title()`
- `plt.xlabel()`
- `plt.ylabel()`

```python
plt.plot(df["Month"], df["Sales"])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

## Adding Legend
Legend help indentify multiple lines in a chart 

```python
plt.plot(df["Month"], df["Sales"],label="Sales")
plt.plot(df["Month"], df["Profit"],label="Profit")

plt.title("Sales vs Profit")
plt.xlabel("Months")
plt.ylabel("Amount")

plt.legend()

plt.show()
```

# 6. Styling Plots

Matplotlib allows customization of:
- Line Style
- Markers
- Line Width

```python
plt.plot(df["Month"], df["Sales"], marker="s", linestyle="--", linewidth=2)

plt.title("Styled Sales Plot")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

# To Enable Grid 

```python
plt.plot(df["Month"], df["Sales"])

plt.grid(True)  # if write false then grid will be removed

plt.show()
```
