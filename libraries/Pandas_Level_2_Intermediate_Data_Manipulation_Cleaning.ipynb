```python

import pandas as pd
import numpy as np

print("Pandas Version:", pd.__version__)
```

## 📊 Creating a Retail Dataset (With Real-World Problems)

```python

data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1005,1005],
    "Customer": ["Alice", "Bob", "Charlie", "David", None, "Eva","Eva"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Phone","Phone"],
    "Quantity": [1, 2, None, 3, 2, 2,2],
    "Price": [70000, 30000, 20000, None, 30000, 30000,30000],
    "City": ["Mumbai", "Delhi", "Mumbai", "Chennai", "Delhi", "Delhi","Delhi"],
    "OrderDate": ["2025-01-01", "2025-01-05", "2025-01-10", "2025-01-12", "2025-01-15", "2025-01-15","2025-01-15"]
}

df = pd.DataFrame(data)
df
```

```python
df.dtypes
```


# 1️⃣ Data Selection & Filtering

## Boolean Indexing


```python

df[df["City"] == "Mumbai"]
```

## Multiple Conditions

```python

df[(df["City"] == "Delhi") & (df["Quantity"] >= 2)]
```

## Using isin()

```python

df[df["City"].isin(["Mumbai", "Chennai"])]
```

## Using query()

```python

df.query("City == 'Delhi' and Quantity >= 2")
```


# 2️⃣ Handling Missing Data


```python

df.isnull().sum()
```

## Filling Missing Values

```python

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].median())
df["Customer"] = df["Customer"].fillna("Unknown")
df
```

## Dropping Missing Values (Alternative)

```python

df.dropna()
```


# 3️⃣ Data Cleaning

## Removing Duplicates


```python

df = df.drop_duplicates()
df
```

## Changing Data Types

```python

df["OrderDate"] = pd.to_datetime(df["OrderDate"])
df.dtypes
df
```

## Creating New Column (TotalAmount)

```python

df["TotalAmount"] = df["Quantity"] * df["Price"]
df
```

## Renaming Columns

```python

df.rename(columns={"City": "Location"}, inplace=True)
df.head()
```


# 4️⃣ GroupBy & Aggregation


```python

df.groupby("Location")["TotalAmount"].sum()
```

## Multiple Aggregations

```python

df.groupby("Product").agg({
    "Quantity": "sum",
    "TotalAmount": ["sum", "mean"]
})
```


# 5️⃣ Merging DataFrames


```python

customer_info = pd.DataFrame({
    "Customer": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Membership": ["Gold", "Silver", "Gold", "Bronze", "Silver"]
})

merged_df = pd.merge(df, customer_info, on="Customer", how="left")
merged_df
```


# 6️⃣ Reshaping Data

## Pivot Table


```python

pivot_table = df.pivot_table(
    values="TotalAmount",
    index="Location",
    columns="Product",
    aggfunc="sum"
)

pivot_table
```
