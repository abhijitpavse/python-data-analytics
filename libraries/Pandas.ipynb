```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
data = {
    "OrderID": [101, 102, 103, 104, 105],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone"],
    "Quantity": [1, 2, 1, 3, 2],
    "Price": [70000, 30000, 20000, 70000, 30000],
    "OrderDate": ["2025-01-01", "2025-01-05", "2025-01-10", "2025-01-12", "2025-01-15"]
}

df = pd.DataFrame(data)
df
```

```python
# creating a series'
quantity_series = pd.Series([1, 2, 3, 4, 5])
quantity_series
```

```python
# series with custom index 
series_with_index = pd.Series([100, 200, 300], index=["A", "B", "C"])
series_with_index
```
